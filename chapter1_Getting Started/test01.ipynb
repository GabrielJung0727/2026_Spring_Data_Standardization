```python
import seaborn as sns
import pandas as pd
import numpy as np

population = sns.load_dataset('tips')

def simple_random_sampling(data, n):
    return data.sample(n=n, random_state=42)

def systematic_sampling(data, step):
    indices = np.arange(0, len(data), step)
    return data.iloc[indices]

def stratified_sampling(data, group_col, n):
    return data.groupby(group_col, group_keys=False).apply(lambda x: x.sample(n=n, random_state=42))

simple_sample = simple_random_sampling(population, 10)  # 단순 임의 추출
systematic_sample = systematic_sampling(population, 20)  # 계통 추출
stratified_sample = stratified_sampling(population, 'day', 2)  # 층화 추출 (요일별 2개씩)

print("Simple Random Sample:")
print(simple_sample)
print("\nSystematic Sample:")
print(systematic_sample)
print("\nStratified Sample:")
print(stratified_sample)
```
